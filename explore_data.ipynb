# Bible Reading Plan - Data Exploration

This notebook demonstrates how to load and view the Excel file headers and data.

```python
from src.config.settings import EXCEL_FILE
from src.services.data_loader import DataLoader
```

## View File Headers

```python
loader = DataLoader(EXCEL_FILE)
headers = loader.get_headers()

print("Excel File Headers:")
print("-" * 40)
for i, header in enumerate(headers):
    print(f"Column {i}: {header}")
```

## Load Chapter Data

```python
chapters = loader.load_chapters()
print(f"Total chapters loaded: {len(chapters)}")
```

## View Sample Data

```python
print("First 10 chapters:")
print("-" * 60)
for chapter in chapters[:10]:
    print(f"{chapter.book_number:2d} | {chapter.book:15s} {chapter.chapter:3d} | {chapter.word_count:4d} words")
```

## Summary Statistics

```python
total_words = sum(chapter.word_count for chapter in chapters)
avg_words = total_words / len(chapters)

print(f"Total words in Bible: {total_words:,}")
print(f"Average words per chapter: {avg_words:.1f}")
print(f"Total chapters: {len(chapters)}")
```

## View as DataFrame (Optional)

```python
import pandas as pd

data = [
    {
        'book_number': ch.book_number,
        'book': ch.book,
        'chapter': ch.chapter,
        'word_count': ch.word_count
    }
    for ch in chapters[:20]
]

df = pd.DataFrame(data)
df
```
