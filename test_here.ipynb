```python
# Import required modules
from src.config.settings import EXCEL_FILE, OUTPUT_DIR, FOUR_YEAR_DAYS
from src.services.data_loader import DataLoader
from src.services.plan_exporter import PlanExporter
from src.strategies.four_year_grouping import FourYearGroupingStrategy

# Load the data
loader = DataLoader(EXCEL_FILE)
chapters = loader.load_chapters()

print(f"Loaded {len(chapters)} chapters from {EXCEL_FILE.name}\n")

# Generate the plan
print("="*70)
print("FOUR-YEAR GROUPING PLAN (12 Groupings per Year)")
print("="*70)

grouping_strategy = FourYearGroupingStrategy(seed=42)
yearly_plans = grouping_strategy.generate_plan(chapters, FOUR_YEAR_DAYS)

# Display in console
PlanExporter.four_year_to_console(yearly_plans)

# Export to files
PlanExporter.four_year_to_csv(yearly_plans, OUTPUT_DIR / "four_year_grouping_plan.csv")
PlanExporter.four_year_to_json(yearly_plans, OUTPUT_DIR / "four_year_grouping_plan.json")

print(f"\n\nFour-year grouping plan exported to {OUTPUT_DIR}")
print(f"  - four_year_grouping_plan.csv")
print(f"  - four_year_grouping_plan.json")
```
